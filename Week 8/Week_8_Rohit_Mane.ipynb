```python
def calculator(expression):
    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return result
    except Exception as e:
        return f"error: {e}"
```

```python
def extract_keywords(text):
    import re
    stopwords = {"the", "is", "in", "at", "of", "a", "an", "and", "or", "to", "for", "on", "with", "this", "that"}
    words = re.findall(r"[a-zA-Z']+", text.lower())
    keywords = [w for w in words if w not in stopwords]
    return list(dict.fromkeys(keywords))
```

```python
def agent(query):
    if not isinstance(query, str) or query.strip() == "":
        return {"type": "error", "result": "empty or invalid query"}

    q = query.lower()

    if "calculate" in q:
        expr = q.replace("calculate", "").strip()
        if expr == "":
            return {"type": "error", "result": "no expression given"}
        return {"type": "calculation", "result": calculator(expr)}

    elif "keywords" in q:
        return {"type": "keywords", "result": extract_keywords(query)}

    else:
        return {"type": "general", "result": query}
```

```python
test_queries = [
    "calculate 12 * (4 + 1)",
    "extract keywords from this sentence about neural networks and data",
    "hello, how are you doing today",
    "calculate",
    ""
]

for tq in test_queries:
    print(agent(tq))
```

```python
while True:
    user_input = input("query> ")
    if user_input.lower() in ("exit", "quit"):
        break
    print(agent(user_input))
```
